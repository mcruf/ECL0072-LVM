```python
#########################
#    Import libraries   #  
#########################
import copernicusmarine
import os
import pandas as pd
import geopandas as gpd


##############################
#    Set working directory   #  
##############################
os.chdir("/Users/mcruf/Library/CloudStorage/OneDrive-Personal/Arbeit/Supervision/ICCAT/")



#############################
#    Set output directory   #  
#############################
# Where the files will be stored
OUTDIR = "/Users/mcruf/OneDrive/Arbeit/Supervision/ICCAT/Dados/GIS/"



##################################
# Import study area bounding box #
##################################
# Will be used to crop rasters to the study area extent.

# Load study area extent
study_area = gpd.read_file("Dados/GIS/Study_area/iccat_extent.shp")

# Extract total bounds: [minx, miny, maxx, maxy]
minx, miny, maxx, maxy = study_area.total_bounds



```

--------------------------

## Global Ocean Physics Reanalysis

- https://data.marine.copernicus.eu/product/GLOBAL_MULTIYEAR_PHY_001_030/description


```python

############################
# Download copernicus data
############################

copernicusmarine.subset(
  dataset_id = "cmems_mod_glo_phy_my_0.083deg_P1D-m", # 'my period' time sieres
  variables = ["thetao", "so", "zos","mlotst"], #sea_water_potential_temperature, salinity, sea surface height, mixed layer thickness
  minimum_longitude=minx,
  maximum_longitude=maxx,
  minimum_latitude=miny,
  maximum_latitude=maxy,
  start_datetime="1993-01-01T00:00:00",
  end_datetime="2016-12-31T00:00:00",
  minimum_depth=0.5,
  maximum_depth=0.5,
  output_filename = "Global_Ocean_Physics_Reanalysis_1993_2016_daily_surface_layers.nc",
  output_directory = OUTDIR
)
```

--------------------------

## Global ocean low and mid trophic levels biomass content hindcast

- https://data.marine.copernicus.eu/product/GLOBAL_MULTIYEAR_BGC_001_033/description


```python
#################################################################
# Download copernicus data - zooplankton & primary productivity
#################################################################

copernicusmarine.subset(
  dataset_id = "cmems_mod_glo_bgc_my_0.083deg-lmtl_P1D-i", # 'my period' time sieres
  variables = [ "zooc", "mnkc_epi", "mnkc_umeso", "npp"], #, zooplankton, epipelagic micronekton,upper mesopelagic micronekton  net primary productivity
  minimum_longitude=minx,
  maximum_longitude=maxx,
  minimum_latitude=miny,
  maximum_latitude=maxy,
  start_datetime="1998-01-01T00:00:00",
  end_datetime="2016-12-31T00:00:00",
  minimum_depth=0.5,
  maximum_depth=0.5,
  output_filename = "Global_ocean_low_and_mid_trophic_levels_biomass_content_hindcast_1998_2016_daily.nc",
  output_directory = OUTDIR
)
```
