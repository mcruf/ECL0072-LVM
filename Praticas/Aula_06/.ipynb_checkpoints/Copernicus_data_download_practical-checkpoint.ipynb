
## Download data from Copernicus Marine Sercive         


- Copernicus provides _physical_, _biological_ and _biochemical_ marine layers at various spatial and temporal resolutions.
- To date (Sep/2026), Copernicus provides > 300 different types of data products: https://data.marine.copernicus.eu/products
- Here, the following data products (variables) will be downloaded:

- **1) Global Ocean Physics Reanalysis**
    - *Spatial resolution*: 0.083 º (~ 8 km)
    - *Temporal resolution*: monhtly
    - *Variables*:
        - SST (sea_water_potential_temperature; thetao)
        - SSH (sea_surface_height_above_geoid; zos)
        - MLT (mixed_layer_thickness; mlotst)
        - BATHY (Bathymetry, deptho)
    - *Download window:* jan 2000 - dec 2001
    - *Products:* (cmems_mod_glo_phy_my_0.083deg_P1M-m , cmems_mod_glo_phy_my_0.083deg_static)

> https://data.marine.copernicus.eu/product/GLOBAL_MULTIYEAR_PHY_001_030/description
> https://documentation.marine.copernicus.eu/PUM/CMEMS-GLO-PUM-001-030.pdf

- **2) Global Ocean Biogeochemistry Hindcast**
    - *Spatial resolution*: 0.25 º (~ 25 km)
    - *Temporal resolution*: monhtly
    - *Variables*:
        - NPPV (net_primary_production; nppv)
    - *Download window:* jan 2000 - dec 2001 
    - *Products:* (cmems_mod_glo_bgc_my_0.25deg_P1M_m)
      
> https://data.marine.copernicus.eu/product/GLOBAL_MULTIYEAR_BGC_001_029/description
> https://documentation.marine.copernicus.eu/PUM/CMEMS-GLO-PUM-001-029.pdf

```python
#~~~~~~~~~~~~~~~~~~~~~~~#
#    Import libraries   #  
#~~~~~~~~~~~~~~~~~~~~~~~#
import copernicusmarine
import os
import pandas as pd
import geopandas as gpd


#~~~~~~~~~~~~~~~~~~~~~~~~~~~~#
#    Set working directory   #  
#~~~~~~~~~~~~~~~~~~~~~~~~~~~~#
os.chdir("/Users/mcruf/Library/CloudStorage/OneDrive-Personal/Arbeit/Lectures_and_Talks/UFRN/Lectures/Graduation/ECL0072-LVM")


#~~~~~~~~~~~~~~~~~~~~~~~~~~~#
#    Set output directory   #  
#~~~~~~~~~~~~~~~~~~~~~~~~~~~#
# Where the files will be stored
OUTDIR = "Dados/GIS/Environmental"


#~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~#
# Import study area bounding box #
#~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~#
# Will be used to crop rasters to the study area extent.

# Load study area extent
study_area = gpd.read_file("Dados/GIS/ICCAT/iccat_atsw.shp")


# Extract total bounds: [minx, miny, maxx, maxy]
minx, miny, maxx, maxy = study_area.total_bounds

```

--------------------------

## Global Ocean Physics Reanalysis


```python

#~~~~~~~~~~~~~~~~~~~#
#  Download  data   #
#~~~~~~~~~~~~~~~~~~~#

## Monthly layer
#~~~~~~~~~~~~~~~~
copernicusmarine.subset(
  dataset_id = "cmems_mod_glo_phy_my_0.083deg_P1M-m", # 'my period' time sieres
  variables = ["thetao", "zos", "mlotst"], #sea_water_potential_temperature, sea surface height, mixed layer thickness
  minimum_longitude=minx,
  maximum_longitude=maxx,
  minimum_latitude=miny,
  maximum_latitude=maxy,
  start_datetime="2000-01-01T00:00:00",
  end_datetime="2001-12-31T00:00:00",
  minimum_depth=0.5,
  maximum_depth=0.5,
  output_filename = "Global_Ocean_Physics_Reanalysis_2000_2001_monthly_surface_layers.nc",
  output_directory = OUTDIR
)
```

```python
## Bathymetry layer
#~~~~~~~~~~~~~~~~~~~~
copernicusmarine.subset(dataset_id = "cmems_mod_glo_phy_my_0.083deg_static", # 'my period' time sieres
                        variables = ["deptho"], #bathymetry
                        minimum_longitude = minx,
                        maximum_longitude = maxx,
                        minimum_latitude = miny,
                        maximum_latitude = maxy,
                        output_filename = "bathymetry.nc",
                        output_directory = OUTDIR)
```

--------------------------

## Global Ocean Biogeochemistry Hindcast



```python
#~~~~~~~~~~~~~~~~~~~#
#  Download  data   #
#~~~~~~~~~~~~~~~~~~~#

copernicusmarine.subset(dataset_id = "cmems_mod_glo_bgc_my_0.25deg_P1M-m", # 'my period' time sieres
                        variables = ["nppv"], # net primary productivity
                        minimum_longitude = minx,
                        maximum_longitude = maxx,
                        minimum_latitude = miny,
                        maximum_latitude = maxy,
                        minimum_depth = 0.51,
                        maximum_depth = 0.51,
                        start_datetime = "2000-01-01T00:00:00",
                        end_datetime = "2001-12-31T00:00:00",
                        output_filename = "Global_Ocean_Biogeochemistry_Hindcast_2000_2001_monthly.nc",
                        output_directory = OUTDIR)
```
